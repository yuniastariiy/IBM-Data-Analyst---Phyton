Next, examine the engagement trends associated with each different claim status.

Start by using Boolean masking to filter the data according to claim status, then calculate the mean and median view counts for each claim status.

```python
claim_videos = data[data['claim_status'] == 'claim']
opinion_videos = data[data['claim_status'] == 'opinion']
```

```python
# Mean and median view count for 'claim' status
claim_mean_views = claim_videos['video_view_count'].mean()
claim_median_views = claim_videos['video_view_count'].median()
```

```python
print(f"Claim - Mean Views: {claim_mean_views:.0f}")
print(f"Claim - Median Views: {claim_median_views:.0f}")
```

```python
# Mean and median view count for 'opinion' status
opinion_mean_views = opinion_videos['video_view_count'].mean()
opinion_median_views = opinion_videos['video_view_count'].median()
```

```python

print(f"Opinion - Mean Views: {opinion_mean_views:.0f}")
print(f"Opinion - Median Views: {opinion_median_views:.0f}")
```

**Question:** What do you notice about the mean and median within each claim category?

**Answer:**  
There is a significant difference in both the mean and median view counts between the two claim categories. Videos labeled as **"claim"** have a much higher average and median view count (mean ≈ 501,029; median ≈ 501,555) compared to videos labeled as **"opinion"** (mean ≈ 4,956; median ≈ 4,953). This suggests that claim videos tend to receive considerably more views than opinion videos. The close proximity between the mean and median within each category also indicates that the distribution of view counts is relatively symmetrical and not heavily skewed within each group.


Now, examine trends associated with the ban status of the author.

Use `groupby()` to calculate how many videos there are for each combination of categories of claim status and author ban status.

```python
# Group by claim_status and author_ban_status, then count
claim_ban_counts = data.groupby(['claim_status', 'author_ban_status']).size().reset_index(name='video_count')
```

```python
# Display result
claim_ban_counts
```

**Question:** What do you notice about the number of claims videos with banned authors? Why might this relationship occur?

**Answer:**  
There are **1,439** claim videos associated with **banned authors**, which is notably higher than the **196** opinion videos associated with banned authors. This suggests that **claim videos are more frequently associated with banned accounts** compared to opinion videos.

This relationship may occur because **claims**—especially those that are misleading, false, or violate community guidelines—are **more likely to be flagged or reported**, leading to **account bans**. In contrast, **opinions** are typically subjective and less likely to breach platform policies, resulting in fewer bans. Therefore, the higher number of bans among authors of claim videos could reflect TikTok’s enforcement of content moderation policies against misinformation or harmful content.


Continue investigating engagement levels, now focusing on `author_ban_status`.

Calculate the median video share count of each author ban status.

```python
# Group by author_ban_status and calculate median of video_share_count
median_share_counts = data.groupby("author_ban_status")["video_share_count"].median()
print(median_share_counts)
```

**Question: What do you notice about the share count of banned authors, compared to that of active authors? Explore this in more depth.**

Videos posted by **banned authors** have a **much higher median share count** (14,468) compared to those posted by **active authors** (437). This suggests that banned authors may have posted content that was more provocative, controversial, or engaging, leading to significantly more shares.

The high share counts could indicate that their videos went viral before being flagged or reviewed. In contrast, active authors have lower median shares, which may reflect more typical or less controversial engagement.

This pattern might imply a relationship between high engagement and content violations, where viral videos from banned accounts may have attracted scrutiny due to potential misinformation or harmful content.


Use `groupby()` to group the data by `author_ban_status`, then use `agg()` to get the count, mean, and median of each of the following columns:
* `video_view_count`
* `video_like_count`
* `video_share_count`

Remember, the argument for the `agg()` function is a dictionary whose keys are columns. The values for each column are a list of the calculations you want to perform.

```python
# Group data by author_ban_status and aggregate count, mean, and median for specified columns
ban_status_summary = data.groupby("author_ban_status")[["video_view_count", "video_like_count", "video_share_count"]].agg(["count", "mean", "median"])
ban_status_summary
```

**Question:** What do you notice about the number of views, likes, and shares for banned authors compared to active authors?

**Answer:**
Videos from **banned authors** tend to receive significantly **higher engagement** than those from **active authors**:

- **Views**: Banned authors have a **mean view count of ~445,845**, compared to **~215,927** for active authors.
- **Likes**: Banned authors average **~153,017 likes**, whereas active authors have only **~71,037 likes** on average.
- **Shares**: Banned authors’ videos have a **median share count of 14,468**, much higher than the **437** for active authors.

This suggests that content from banned authors might be more **viral or controversial**, leading to greater engagement. However, it may also indicate that high engagement could correlate with content that violates community guidelines, resulting in bans.

Now, create three new columns to help better understand engagement rates:
* `likes_per_view`: represents the number of likes divided by the number of views for each video
* `comments_per_view`: represents the number of comments divided by the number of views for each video
* `shares_per_view`: represents the number of shares divided by the number of views for each video

```python
# Create new columns for engagement rates
data['likes_per_view'] = data['video_like_count'] / data['video_view_count']
data['comments_per_view'] = data['video_comment_count'] / data['video_view_count']
data['shares_per_view'] = data['video_share_count'] / data['video_view_count']

# Display the first few rows to confirm the new columns are added
data[['video_view_count', 'video_like_count', 'likes_per_view',
      'video_comment_count', 'comments_per_view',
      'video_share_count', 'shares_per_view']].head()
```

Use `groupby()` to compile the information in each of the three newly created columns for each combination of categories of claim status and author ban status, then use `agg()` to calculate the count, the mean, and the median of each group.

```python
# Group by claim_status and author_ban_status, then aggregate the new engagement rate columns
engagement_summary = data.groupby(['claim_status', 'author_ban_status'])[
    ['likes_per_view', 'comments_per_view', 'shares_per_view']
].agg(['count', 'mean', 'median'])

# Display the result
engagement_summary
```

**Question: How does the data for claim videos and opinion videos compare or differ?**

When comparing **claim** videos and **opinion** videos across views, comments, likes, and shares, several key differences emerge:

- **Likes per view**: Claim videos tend to have a **higher like-to-view ratio** (around 0.33–0.35) compared to opinion videos (around 0.21–0.23). This suggests that viewers are more likely to like content that is framed as a claim rather than an opinion.

- **Comments per view**: Although the overall values are small, claim videos consistently receive **more comments per view** (0.00137–0.00139) than opinion videos (0.00043–0.00054), indicating greater viewer engagement and possibly more discussion or debate in response to claims.

- **Shares per view**: Claim videos also have a **higher share rate** (0.065–0.068) compared to opinion videos (0.041–0.045). This implies that claims may be perceived as more compelling or important to spread.

**Summary**: Across all metrics likes, comments, and shares **claim videos consistently show higher engagement per view than opinion videos**. This pattern could be important for modeling and content strategy, as it may reflect how users interact differently with factual or assertive content versus personal opinions.
