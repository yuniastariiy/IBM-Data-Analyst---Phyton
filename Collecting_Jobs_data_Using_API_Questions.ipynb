## Lab: Collect Jobs Data using Jobs API


### Objective: Determine the number of jobs currently open for various technologies  and for various locations


```python
#Import required libraries
import pandas as pd
import json
import requests
```

```python
api_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"

def get_number_of_jobs_T(technology):
    response = requests.get(api_url)
    data = response.json()

    # Filter berdasarkan 'Job Title'
    filtered_jobs = [job for job in data if technology.lower() in job.get('Job Title', '').lower()]

    number_of_jobs = len(filtered_jobs)
    return technology, number_of_jobs
```

```python
get_number_of_jobs_T("Python")
```

#### Write a function to find number of jobs in US for a location of your choice


```python
def get_number_of_jobs_L(location):
    response = requests.get(api_url)
    data = response.json()

    filtered_jobs = [job for job in data if location.lower() in job.get('Location', '').lower()]

    number_of_jobs = len(filtered_jobs)
    return location, number_of_jobs
```

### Store the results in an excel file


Collect the number of job postings for the following locations using the API:

* Los Angeles
* New York
* San Francisco
* Washington DC
* Seattle
* Austin
* Detroit


```python
# Fetch data from the API
response = requests.get(api_url)
data = response.json()

# Convert JSON data to DataFrame
df = pd.DataFrame(data)
```

```python
# Define list of locations to search
locations = ['Los Angeles', 'New York', 'San Francisco', 'Washington DC', 'Seattle', 'Austin', 'Detroit']
```

```python
# Function to count job postings for a specific location
def get_number_of_jobs_L(location):
    count = df['Location'].str.contains(location, case=False, na=False).sum()
    return location, count

# Create Excel workbook and worksheet
wb = Workbook()
ws = wb.active
ws.append(['Location', 'Number of Jobs'])  # Header row

# Collect results for each location and write to Excel
results = []
for loc in locations:
    loc_name, job_count = get_number_of_jobs_L(loc)
    ws.append([loc_name, job_count])
    results.append((loc_name, job_count))

# Save to Excel file
wb.save("location_job_counts.xlsx")
```

```python
# Display result in Colab as DataFrame
result_df = pd.DataFrame(results, columns=['Location', 'Number of Jobs'])
result_df
```

Collect the number of job postings for the following languages using the API:

```python
# Define the list of technologies to search for
technologies = [
    'C', 'C#', 'C++', 'Java', 'JavaScript', 'Python',
    'Scala', 'Oracle', 'SQL Server', 'MySQL Server', 'PostgreSQL', 'MongoDB'
]
```

```python
# Function to count job postings that mention a specific technology
def get_number_of_jobs_T(technology):
    count = df['Key Skills'].str.contains(technology, case=False, na=False).sum()
    return technology, count

# Create workbook
wb = Workbook()
ws = wb.active
ws.append(['Technology', 'Number of Jobs'])

# Loop through each technology and collect job counts
results = []
for tech in technologies:
    tech_name, job_count = get_number_of_jobs_T(tech)
    ws.append([tech_name, job_count])
    results.append((tech_name, job_count))

# Save to Excel
wb.save("tech_job_counts.xlsx")
```
